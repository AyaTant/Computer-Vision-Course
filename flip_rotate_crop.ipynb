```python
import cv2
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("./mountain.jpg")
```

```python
img_rgb = cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
```

```python
# 1 = horixontally
# 2 = vertically
# -1 = both axes
flip_horizontal = cv2.flip(image,1)
flip_vertical = cv2.flip(image, 2)
flip_both = cv2.flip(image,-1)

titles = ["Original Image","Horizontal Image","Vertical Image","Flip Both Axes"]
images = [
    img_rgb,
    flip_horizontal,
    flip_vertical,
    flip_both
]
plt.figure(figsize=(12,8))
for i in range(4):
    plt.subplot(2,2,i+1)
    plt.imshow(images[i])
    plt.title(titles[i])
    plt.axis("Off")

plt.tight_layout()
plt.show()
```

# Rotate

```python
rotate_90_clockwise = cv2.rotate(img_rgb, cv2.ROTATE_90_CLOCKWISE)
rotate_90_counterclockwise = cv2.rotate(img_rgb, cv2.ROTATE_90_COUNTERCLOCKWISE)
rotate_180 = cv2.rotate(img_rgb, cv2.ROTATE_180)

titles = ["Original Image","90 Clockwise Image","90 Counter Clockwise Image","180 Image"]
images = [
    img_rgb,
    rotate_90_clockwise,
    rotate_90_counterclockwise,
    rotate_180
]
plt.figure(figsize=(12,8))
for i in range(4):
    plt.subplot(2,2,i+1)
    plt.imshow(images[i])
    plt.title(titles[i])
    plt.axis("Off")

plt.tight_layout()
plt.show()
```

# Cropping an Image

```python
print(img_rgb.shape)
```

```python
x_start, y_start = 400, 300
x_end, y_end = 800,900

cropped_img = img_rgb[y_start:y_end, x_start:x_end]

plt.figure(figsize=(8,4))
plt.subplot(1,2,1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("Off")

plt.subplot(1,2,2)
plt.imshow(cropped_img)
plt.title("Cropped Image")
plt.axis("Off")

plt.tight_layout()
plt.show()
```
