```python
import cv2
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("./mountain.jpg")
```

```python
img_rgb = cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(img_rgb)
```

## Image resizing

```python
print(img_rgb.shape)
```

```python
new_height, new_width = 400,400

resized_img = cv2.resize(img_rgb, (new_width,new_height))

plt.imshow(resized_img)
```

## Image Scaling

```python
scale_percentage = 50
s_width = int(img_rgb.shape[1] * scale_percentage /100)
s_height = int(img_rgb.shape[0] * scale_percentage /100)

scaled_img = cv2.resize(img_rgb, (s_width,s_height), interpolation=cv2.INTER_AREA)
print(scaled_img.shape)
plt.imshow(scaled_img)
plt.title("Scaled Image")
```

## Interpolation

```python
# img_rgb = cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
resize_width = 600
resize_height = 600

# interpolation methods
methods ={
    'Nearest' : cv2.INTER_NEAREST,
    'Linear' : cv2.INTER_LINEAR,
    'Cubic': cv2.INTER_CUBIC,
    'Area': cv2.INTER_AREA,
    'Lancsoz4': cv2.INTER_LANCZOS4
}
resized_imgs = {}
for name,method in methods:
    resized_imgs[name] = cv2.resize(img_rgb, (resize_width,resize_height),interpolation=method)
    cv2.imwrite(f"./{name}.jpg",resized_imgs[name])

plt.figure(figsize=(12,8))
# original image
plt.subplot(2,3,1)
plt.imshow(img_rgb)
plt.title("Original")
plt.axis('off')

for i, (name,resized_img) in enumerate(resized_imgs.items(),start=2):
    plt.subplot(2,3,i)
    plt.imshow(resized_img)
    plt.title(name)
    plt.axis('off')

plt.tight_layout()
plt.show()
```
