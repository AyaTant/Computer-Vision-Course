```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
bgr_image = cv2.imread("./mountain.jpg")
```

```python
# convert to grayscale
gray_img = cv2.cvtColor(bgr_image, cv2.COLOR_BGR2GRAY)
cv2.imwrite("./gray_image.jpg",gray_img)
```

```python
# convert to HSV
hsv_img = cv2.cvtColor(bgr_image, cv2.COLOR_BGR2HSV)
cv2.imwrite("./hsv_image.jpg",hsv_img)
```

```python
lab_img = cv2.cvtColor(bgr_image, cv2.COLOR_BGR2LAB)
cv2.imwrite("./lab_image.jpg",lab_img)
```

```python
# convert to RGB
rgb_img = cv2.cvtColor(bgr_image, cv2.COLOR_BGR2RGB)
cv2.imwrite("./rgb_image.jpg",rgb_img)
```

```python
cv2.imwrite("./rgb_image.jpg",bgr_image)
```

```python
image = cv2.imread("./mountain.jpg")
```

```python
img_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
img_hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
```

```python
plt.imshow(image)
```

```python
lower_white = np.array([0,0,190])
upper_white = np.array([30,80,255])
```

```python
mask = cv2.inRange(img_hsv,lower_white,upper_white)
```

```python
plt.imshow(mask)
```

```python
black_hsv = np.uint8([[[120,255,0]]])
black_bgr = cv2.cvtColor(black_hsv,cv2.COLOR_HSV2BGR)[0][0]
```

```python
image_result = image.copy()
image_result[mask>0] = black_bgr
```

```python
img_result_rgb = cv2.cvtColor(image_result,cv2.COLOR_BGR2RGB)
```

```python
# plot
fig, axs = plt.subplots(1,3,figsize=(12,6))
axs[0].imshow(img_rgb)
axs[0].set_title("Original Image")
axs[0].axis("off")

axs[1].imshow(mask)
axs[1].set_title("Binary Mask")
axs[1].axis("off")

axs[2].imshow(img_result_rgb)
axs[2].set_title("Output Image")
axs[2].axis("off")

plt.tight_layout()
plt.show()
```
